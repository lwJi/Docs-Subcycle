# Plots for Slides

```julia
using Plots
using LaTeXStrings
using Measures
theme(:wong2)
```

```julia
ratioToctbypat(L) = 2.0^(L-1) / (2.0^L-1)
ratioZoctbypat(L) = 2.0^(L-4) * (1+7L) / (2.0^L-1)
```

```julia
let levs = 1:15
    xs = 1:3:14
    #
    plt1 = plot(tickfontsize=10, legendfontsize=8, framestyle=:box)
    xticks!(plt1, xs, [L"%$(n)" for n in xs])
    ys = 0:0.2:1.0
    yticks!(plt1, ys, [L"%$(n)" for n in ys])
    plt1 = scatter!(levs, ratioToctbypat.(levs), label="", linewidth=2)
    plt1 = plot!(levs, [0.5 for i in 1:length(levs)], linecolor=:gray, linestyle=:dash, label="")
    plt1 = plot!(xlim=(1-0.2,15+0.2),ylim=(0,1.03),
                 #xlabel=L"\mathrm{Number\,\,of\,\,levels}",
                 ylabel=L"T^\mathrm{octree}/T^\mathrm{patch}")

    #
    plt2 = plot(tickfontsize=10, legendfontsize=8, framestyle=:box)
    xticks!(plt2, xs, [L"%$(n)" for n in xs])
    ys = 2:2:6
    yticks!(plt2, ys, [L"%$(n)" for n in ys])
    plt2 = scatter!(levs, ratioZoctbypat.(levs), label="", linewidth=2)
    plt2 = plot!(levs, [1 for i in 1:length(levs)], linecolor=:gray, linestyle=:dash, label="")
    plt2 = plot!(xlim=(1-0.2,15+0.2), ylim=(0,7),
                 xlabel=L"\mathrm{Number\,\,of\,\,levels}",
                 ylabel=L"Z^\mathrm{octree}/Z^\mathrm{patch}")

    plt = plot(plt1, plt2, layout=(2,1), size=(400, 400), dpi=60)
    # savefig(plt, "ratios.pdf")
    display(plt)
end
```
